# Resumo Didático: Capítulo 3 – Regressão Linear (ISLP)

A **Regressão Linear** pode ser compreendida através de uma jornada lógica dividida em quatro passos:
1. **Como traçar a linha?** (O modelo)
2. **Essa linha funciona?** (Métricas e testes)
3. **Como lidar com o mundo real?** (Features categóricas e relações curvas)
4. **O que pode dar errado?** (Diagnóstico dos resíduos)

---

## 1. A Ideia Central: A Corrida da Linha Reta

Imagine prever o volume de vendas de um produto ($Y$) com base no investimento em anúncios de TV ($X$):

$$Y \approx \beta_0 + \beta_1 X$$

* **$\beta_0$ (Intercepto / Ponto de Partida):** O valor esperado de vendas caso o investimento em TV seja zero.
* **$\beta_1$ (Inclinação / Peso):** A cada $1$ real investido a mais em TV, quantas vendas adicionais entram no caixa.

### Como o Computador Encontra Essa Reta? (Mínimos Quadrados / OLS)
O algoritmo tenta posicionar uma reta entre os pontos observados:
* A distância vertical entre o ponto real e a reta traçada é o **resíduo (erro)**.
* Para evitar que erros positivos e negativos se anulem, o algoritmo eleva todos os erros ao quadrado e soma tudo (**RSS – Residual Sum of Squares**).
* A reta escolhida é aquela que resulta na **menor soma de erros quadráticos possível**.

---

## 2. A Linha Funciona? (Os 4 Testes de Qualidade)

Após traçar a reta, utilizam-se métricas e testes para verificar a validade do aprendizado:

1. **O $p$-valor e o Teste $t$ (Esse preditor realmente importa?):**
   * **Hipótese nula ($H_0$):** O investimento em TV não tem relação com as vendas ($\beta_1 = 0$).
   * Se o $p\text{-valor} < 0{,}05$, rejeita-se a hipótese nula, indicando que a relação observada não é fruto do acaso.
2. **O Teste $F$ (O conjunto todo tem sinal estatístico?):**
   * Em modelos com múltiplos preditores (ex.: TV, Rádio, Jornal), o Teste $F$ avalia se **pelo menos uma** das variáveis auxilia na previsão. Analisar apenas os $p$-valores individuais pode gerar conclusões equivocadas devido ao problema de testes múltiplos.
3. **$R^2$ (Quanto da variação é explicada?):**
   * Varia de $0$ a $1$ (ou $0\%$ a $100\%$).
   * Um $R^2 = 0{,}80$ significa que $80\%$ da variabilidade das vendas é explicada pelos gastos em anúncios. Os $20\%$ restantes correspondem a fatores não capturados pelo modelo.
4. **RSE – Residual Standard Error (Qual a magnitude do erro médio?):**
   * Mede o desvio médio na própria escala da variável resposta (ex.: o modelo erra, em média, por cerca de $3{,}2$ mil unidades).

---

## 3. Adaptando o Modelo ao Mundo Real

Na prática, as relações raramente são puramente lineares e contínuas:

* **Variáveis Categóricas (Ex.: Comprou Casa: Sim / Não):**
  * O computador cria variáveis binárias (*dummy* ou *one-hot encoding*), atribuindo $1$ para Sim e $0$ para Não.
  * Quando a observação recebe $0$, essa parcela é zerada, restando apenas o valor de referência capturado no intercepto $\beta_0$.
* **Efeito Sinérgico / Interação ($X_1 \times X_2$):**
  * Muitas vezes, investir simultaneamente em TV e Rádio gera um impacto superior à soma de cada canal isolado. Adiciona-se o termo multiplicativo $(\text{TV} \times \text{Rádio})$ para modelar esse efeito conjunto.
  * **Princípio Hierárquico:** Ao incluir a interação, mantêm-se os efeitos principais individuais no modelo, mesmo que seus $p$-valores isolados sejam altos.
* **Relações Curvas (Regressão Polinomial):**
  * Para fenômenos não lineares (como a relação entre potência do motor e consumo), podem-se adicionar termos elevados a potências ($X^2$, $X^3$) ou logaritmos ($\log X$). O modelo continua sendo linear porque a linearidade reside nos parâmetros $\beta$, e não nas features.

---

## 4. O Diagnóstico dos Resíduos

Para garantir a confiabilidade do modelo em produção, os resíduos (erros) devem se comportar como ruído puramente aleatório. Padrões visuais nos resíduos indicam violações de premissas:

| Padrão nos Gráficos de Resíduos | Significado Prático | Como Corrigir |
| :--- | :--- | :--- |
| **Padrão em "U"** | Relação subjacente é não linear (uma reta foi usada onde há curvatura). | Adicionar transformações polinomiais ($X^2$) ou não lineares ($\log X$). |
| **Formato de Funil (Heterocedasticidade)** | A variância dos erros não é constante (o modelo erra mais à medida que a resposta aumenta). | Aplicar transformações côncavas na variável resposta, como $\log(Y)$ ou $\sqrt{Y}$. |
| **Outliers Isolados** | Pontos atípicos com resíduos muito distantes da média que distorcem as métricas gerais. | Investigar falhas de coleta ou erros de registro; tratar ou descartar com justificativa. |
| **Alta Alavancagem (*High Leverage*)** | Observações com valores extremos nas variáveis $X$ que puxam a reta em sua direção. | Avaliar o impacto da observação comparando o ajuste com e sem ela. |
| **Multicolinearidade (VIF elevado)** | Variáveis preditoras altamente correlacionadas competindo entre si, gerando instabilidade nos coeficientes. | Remover variáveis redundantes ou aplicar técnicas de regularização (como Ridge ou Lasso). |

---

## 5. Resumo para Guardar

1. A **Regressão Linear** busca os coeficientes que minimizam a soma dos resíduos quadráticos (RSS) entre os valores reais e as predições.
2. Um **$R^2$ elevado** deve sempre ser acompanhado da análise do Teste $F$, dos $p$-valores e da verificação de multicolinearidade.
3. O **gráfico de resíduos** serve como ferramenta diagnóstica: ausência de padrões indica bom comportamento estocástico; padrões estruturados exigem transformações nos dados ou ajuste na forma funcional.

# Resumo Prático: Capítulo 3 – Regressão Linear (ISLP)

O Capítulo 3 aborda o método supervisionado mais clássico e interpretável do aprendizado estatístico. Ele detalha a mecânica dos Mínimos Quadrados Ordinários (OLS), os testes de hipótese para seleção de features, métricas de qualidade de ajuste, extensões não lineares e os principais problemas em dados reais (multicolinearidade, heterocedasticidade, outliers e alta alavancagem).

---

## 1. Regressão Linear Simples

Modela a relação entre um único preditor quantitativo $X$ e uma resposta contínua $Y$ através de uma linha reta:

$$Y \approx \beta_0 + \beta_1 X$$


* $\beta_0$: Intercepto (valor esperado de $Y$ quando $X = 0$).
* $\beta_1$: Coeficiente angular ou inclinação (*slope* — impacto médio em $Y$ decorrente de uma unidade adicional em $X$).

### A. Estimando os Coeficientes (Mínimos Quadrados / OLS)
O método de mínimos quadrados minimiza a Soma dos Resíduos Quadráticos ($RSS$):

$$RSS = \sum_{i=1}^n e_i^2 = \sum_{i=1}^n (y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i)^2$$


As estimativas analíticas fechadas que minimizam o $RSS$ são:

$$\hat{\beta}_1 = \frac{\sum_{i=1}^n (x_i - \bar{x})(y_i - \bar{y})}{\sum_{i=1}^n (x_i - \bar{x})^2} \quad \text{e} \quad \hat{\beta}_0 = \bar{y} - \hat{\beta}_1 \bar{x}$$


> **Propriedade Matemática Fundamental:** A reta ajustada por OLS sempre passa pelo centroide das médias $(\bar{x}, \bar{y})$.

### B. Avaliando a Precisão dos Coeficientes
* **Erro Padrão ($SE$):** Mede o desvio da estimativa do coeficiente em relação à média populacional.
* **Intervalo de Confiança de 95%:** Define a faixa que conterá o verdadeiro valor populacional em 95% das replicações:
  $$\hat{\beta}_1 \pm 2 \cdot SE(\hat{\beta}_1)$$

* **Teste de Hipóteses:**
  * $H_0: \beta_1 = 0$ (Não há relação entre $X$ e $Y$).
  * $H_a: \beta_1 \neq 0$ (Existe relação entre $X$ e $Y$).
  * Estatística $t$:
    $$t = \frac{\hat{\beta}_1 - 0}{SE(\hat{\beta}_1)}$$

  * Se o $p\text{-valor}$ for suficientemente pequeno (tipicamente $< 0.05$ ou $< 0.01$), rejeita-se $H_0$ e conclui-se que o preditor é estatisticamente significante.

### C. Avaliando a Qualidade do Ajuste do Modelo
1. **Erro Padrão Residual ($RSE$):** Estimativa do desvio padrão do erro irredutível $\epsilon$, calculado em unidades da própria variável resposta:
   $$RSE = \sqrt{\frac{1}{n - 2} RSS}$$

2. **Estatística $R^2$:** Proporção da variância total da resposta que é explicada pelo modelo:
   $$R^2 = 1 - \frac{RSS}{TSS} = \frac{TSS - RSS}{TSS}$$
   *(onde $TSS = \sum (y_i - \bar{y})^2$ é a soma total dos quadrados)*.
   * Em regressão simples, $R^2 = \text{Cor}(X, Y)^2$.

---

## 2. Regressão Linear Múltipla

Acomoda $p$ preditores simultaneamente em uma única equação:

$$Y = \beta_0 + \beta_1 X_1 + \beta_2 X_2 + \dots + \beta_p X_p + \epsilon$$


* **Interpretação de $\beta_j$:** O efeito médio na resposta $Y$ ao aumentar $X_j$ em 1 unidade, **mantendo todos os outros preditores constantes** (*ceteris paribus*).

### As Perguntas Críticas da Regressão Múltipla:

#### 1. Pelo menos um preditor é útil? (Teste $F$ Global)
Não se deve olhar os $p\text{-valores}$ individuais dos testes $t$ quando há muitas features (risco de falsos positivos pelo problema de múltiplos testes). Usa-se a **estatística $F$**:

* $H_0: \beta_1 = \beta_2 = \dots = \beta_p = 0$
* $H_a: \text{Pelo menos um } \beta_j \neq 0$
$$F = \frac{(TSS - RSS)/p}{RSS/(n - p - 1)}$$

Se $F \gg 1$ e o $p\text{-valor}$ do teste $F$ for próximo de zero, conclui-se que o modelo tem sinal estatístico relevante.

#### 2. Como testar subconjuntos de variáveis? (Teste $F$ Parcial)
Para verificar se um grupo específico de $q$ variáveis pode ser removido do modelo sem perda de poder explicativo:

$$F = \frac{(RSS_0 - RSS)/q}{RSS/(n - p - 1)}$$


*(onde $RSS_0$ é o erro residual do modelo reduzido sem as $q$ variáveis)*.

---

## 3. Considerações Adicionais na Modelagem Linear

### A. Preditores Qualitativos / Categóricos
* **2 Categorias (Binária):** Cria-se 1 variável indicadora (*dummy*) com valores $\{0, 1\}$.
* **$K$ Categorias:** Criam-se sempre $K - 1$ variáveis *dummy*. A categoria omitida torna-se o **nível de referência (*baseline*)**, sendo capturada diretamente pelo intercepto $\beta_0$.

### B. Extensões do Modelo Linear

#### 1. Termos de Interação (*Synergy / Effect Modification*)
O modelo aditivo padrão assume que o efeito de $X_1$ em $Y$ independe do valor de $X_2$. A inclusão do termo multiplicativo relaxa essa premissa:

$$Y = \beta_0 + \beta_1 X_1 + \beta_2 X_2 + \beta_3 (X_1 X_2) + \epsilon = \beta_0 + (\beta_1 + \beta_3 X_2) X_1 + \beta_2 X_2 + \epsilon$$


> **Princípio Hierárquico (*Hierarchical Principle*):** Se incluirmos um termo de interação ($X_1 X_2$) no modelo, **devemos manter os efeitos principais ($X_1$ e $X_2$)**, mesmo que seus $p\text{-valores}$ individuais não sejam estatisticamente significantes.

#### 2. Relações Não Lineares (Regressão Polinomial)
Transformações não lineares (como $X^2, X^3, \sqrt{X}, \log X$) podem ser incluídas como novas colunas na matriz de features sem deixar de ser um modelo linear em relação aos parâmetros $\beta$:

$$\text{mpg} = \beta_0 + \beta_1 \cdot \text{horsepower} + \beta_2 \cdot \text{horsepower}^2 + \epsilon$$

## 4. Problemas Potenciais e Diagnóstico de Resíduos

Em produção e validação estatística, seis problemas clássicos invalidam as premissas do modelo OLS:

| Problema | Sintoma Visual / Métrico | Risco / Consequência | Ação Corretiva |
| :--- | :--- | :--- | :--- |
| **1. Não Linearidade** | Padrão em forma de "U" no gráfico de Resíduos vs. Valores Ajustados. | Previsões sistematicamente enviesadas nas pontas. | Aplicar transformações não lineares nos preditores ($\log X, \sqrt{X}, X^2$). |
| **2. Correlação dos Erros** | Resíduos adjacentes com valores similares no tempo (*tracking*). | Subestima os erros-padrão reais ($SE$), inflando a confiança estatística. | Modelagem de séries temporais, modelos autorregressivos (ARIMA). |
| **3. Heterocedasticidade** | Gráfico de resíduos em formato de funil (*funnel shape* — variância cresce com o alvo). | Intervalos de predição e testes de hipótese inválidos. | Transformação côncava na variável resposta (ex: $\log(Y)$ ou $\sqrt{Y}$). |
| **4. Outliers** | Resíduo studentizado $> 3$ ou $< -3$. | Distorce o $RSE$, reduz o $R^2$ e infla os erros-padrão. | Investigar falhas de coleta; remover com critério justificado. |
| **5. Alta Alavancagem (*High Leverage*)** | Ponto com valores incomuns nas features $X$ (estatística de alavancagem $h_i$ elevada). | Puxa a reta de regressão para perto de si, mascarando o erro residual. | Inspecionar a observação e avaliar o ajuste com e sem ela. |
| **6. Colinearidade** | $VIF > 5$ ou $10$ (*Variance Inflation Factor*). | Reduz o poder estatístico; torna os coeficientes instáveis e com $SE$ gigante. | Dropar uma das variáveis redundantes ou combiná-las em um índice sintético. |

---

## 5. Regressão Linear vs. KNN (Paramétrico vs. Não Paramétrico)

* **Em 1 dimensão ($p = 1$):** O KNN pode superar o OLS se a relação real de $f$ for altamente curvilínea ou irregular.
* **Em altas dimensões ($p \ge 4$):** O modelo linear quase sempre supera o KNN devido à **Maldição da Dimensionalidade (*Curse of Dimensionality*)**.
  * Conforme $p$ cresce, as observações tornam-se esparsas no espaço multidimensional, e os vizinhos mais próximos deixam de ser "próximos", degradando drasticamente o MSE de teste do KNN.

## 6. O Laboratório em Python do Capítulo 3 (`Lab 3.6`)

O laboratório utiliza os pacotes `statsmodels` e o helper `ISLP` para ajustar modelos e gerar resumos estatísticos:

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
from ISLP import load_data
from ISLP.models import ModelSpec as MS, summarize, poly
from statsmodels.stats.outliers_influence import variance_inflation_factor as VIF

# Carregamento do dataset Boston
Boston = load_data('Boston')

# 1. Regressão Linear Simples
X = MS(['lstat']).fit_transform(Boston)
y = Boston['medv']
model = sm.OLS(y, X)
results = model.fit()
print(summarize(results))

# 2. Regressão Múltipla com todas as variáveis
features = Boston.columns.drop('medv')
X_all = MS(features).fit_transform(Boston)
model_all = sm.OLS(y, X_all).fit()

# 3. Cálculo do VIF (Multicolinearidade)
vals = [VIF(X_all.values, i) for i in range(1, X_all.shape[1])]
vif_df = pd.DataFrame({'vif': vals}, index=X_all.columns[1:])

# 4. Termos de Interação
X_inter = MS(['lstat', 'age', ('lstat', 'age')]).fit_transform(Boston)
model_inter = sm.OLS(y, X_inter).fit()

# 5. Transformações Polinomiais (Grau 2)
X_poly = MS([poly('lstat', degree=2), 'age']).fit_transform(Boston)
model_poly = sm.OLS(y, X_poly).fit()
```

## 7. Checklist Mental para Entrevistas e Produção (Capítulo 3)

1. **Olhe o Teste $F$ antes do Teste $t$:** Nunca conclua relevância olhando apenas para $p\text{-valores}$ individuais se $p$ for grande; confirme primeiro se o modelo como um todo rejeita $H_0$ via estatística $F$.
2. **Respeite o Princípio Hierárquico:** Ao testar interações ($X_1 \cdot X_2$), nunca descarte os termos individuais $X_1$ ou $X_2$ por terem $p\text{-valor}$ alto se o termo de interação for significante.
3. **Cuidado com a Colinearidade silenciosa:** Coeficientes com sinais invertidos ou magnitudes estranhas são o principal sintoma prático de VIF elevado entre as features.
4. **Resíduos revelam a verdade:** Antes de entregar qualquer baseline OLS para produção, valide o gráfico de Resíduos vs. Valores Ajustados para descartar heterocedasticidade e não linearidade.
