```python
import numpy as np
import pandas as pd
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingRegressor
import matplotlib.pyplot as plt
```

```python
# Carregando o dataset California Housing nativo do sklearn
california = fetch_california_housing(as_frame=True)
X = california.data
y = california.target
```

```python
# Dividindo em dados de treino (70%) e teste (30%)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Criando e treinando o Gradient Boosting (com 5000 árvores rasas)
boost_modelo = GradientBoostingRegressor(n_estimators=5000, learning_rate=0.001, max_depth=3, random_state=0)
boost_modelo.fit(X_train, y_train)
```

```python
# Calculando o MSE de Teste
previsoes_boost = boost_modelo.predict(X_test)
mse_boost = np.mean((y_test - previsoes_boost)**2)
print(f"MSE do Gradient Boosting: {mse_boost:.2f}")
```

```python
# 1. Extrair as importâncias do Random Forest e os nomes das variáveis
importancias = boost_modelo.feature_importances_
nomes_variaveis = X.columns 

# 2. Criar um DataFrame e ordenar os valores do menor para o maior
df_importancias = pd.DataFrame({'Importância': importancias}, index=nomes_variaveis)
df_importancias = df_importancias.sort_values(by='Importância', ascending=True)

# 3. Gerar o gráfico de barras horizontal
df_importancias.plot(kind='barh', figsize=(8, 6), color='skyblue', legend=False)
plt.title('Importância das Variáveis - Random Forest')
plt.xlabel('Nível de Importância')
plt.ylabel('Variáveis')
plt.show()
```
