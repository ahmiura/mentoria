A notação $O(n)$ (lê-se "Big-O de n" ou "Grande O de n") representa a complexidade de um algoritmo ou cálculo.

No contexto de computação, matemática e ciência de dados, essa notação descreve como o esforço computacional cresce à medida que o tamanho dos dados aumenta.

O que significa cada parte?
* O ($O$ maiúsculo): Vem de "Order of" (Ordem de magnitude). Ele indica o limite superior do tempo de execução ou do uso de memória quando os dados ficam grandes.
* A variável entre parênteses: É a dimensão dos dados que está crescendo:
  * Na literatura estatística e de Machine Learning (como no livro ISLP), utiliza-se $n$ para o número de observações/amostras e $p$ para o número de preditores/variáveis.   

1. $O(1)$
* Como se lê: "Ó de um" (ou "Grande Ó de um").
* Nome: Complexidade Constante.
* Significado: O tempo de execução não depende do tamanho dos dados de entrada. Seja com 10 ou 1 milhão de itens, a operação leva o mesmo tempo (por exemplo: acessar o primeiro elemento de um vetor pelo índice).

2. $O(\log n)$
* Como se lê: "Ó de log de n" (ou "Grande Ó de log de n").
* Nome: Complexidade Logarítmica.
* Significado: O tempo cresce de forma muito lenta à medida que os dados aumentam. Geralmente ocorre em algoritmos que dividem o volume de dados pela metade a cada passo (como a busca binária).

3. $O(n)$
* Como se lê: "Ó de n" (ou "Grande Ó de n").
* Nome: Complexidade Linear.
* Significado: O tempo cresce em proporção direta ao tamanho da entrada. Se os dados dobram de tamanho, o número de passos também dobra (por exemplo: um laço for simples percorrendo uma lista inteira).

4. $O(n \log n)$
* Como se lê: "Ó de n log de n".
* Nome: Complexidade Linear-logarítmica (ou "Quase linear" / "Log-linear").
* Significado: Um crescimento um pouco superior ao linear, mas ainda muito eficiente. É a complexidade típica dos algoritmos clássicos e eficientes de ordenação (como Merge Sort e Quick Sort em caso médio).

5. $O(n^2)$
* Como se lê: "Ó de n ao quadrado" (ou "Grande Ó de n dois").
* Nome: Complexidade Quadrática.
* Significado: O tempo cresce ao quadrado do tamanho da entrada. Se os dados dobram de tamanho, o esforço multiplica por $4$ ($2^2$). Ocorre frequentemente em laços aninhados (um for dentro de outro for comparando todos os pares de itens).

6. $O(p^3)$
* Como se lê: "Ó de p ao cubo" (ou "Grande Ó de p três").
* Nome: Complexidade Cúbica.
* Significado: O tempo cresce na terceira potência em relação a $p$ (número de atributos/features). Se $p$ dobrar, o esforço computacional é multiplicado por $8$ ($2^3$). Na regressão linear, é o custo exigido para inverter matrizes do tipo $(p \times p)$ nas equações normais.

7. $O(2^n)$
* Como se lê: "Ó de dois elevado a n".
* Nome: Complexidade Exponencial.
* Significado: O tempo dobra a cada elemento a mais que entra na base de dados. Torna-se inviável de ser executado para valores até mesmo pequenos de $n$ (como algoritmos de força bruta que tentam testar todos os subconjuntos possíveis de um grupo).

### 1. Complexidade de Treino vs. Inferência nos Modelos de ML

Um modelo pode ter treino lento, mas inferência instantânea (ou o oposto). Em uma sabatina, espera-se que você saiba de cabeça ou deduza as complexidades dos principais algoritmos:

| Algoritmo | Treino (Tempo) | Inferência (Tempo) | Gargalo de Produção |
| :--- | :--- | :--- | :--- |
| **Regressão Linear (OLS - Equações Normais)** | $O(n \cdot p^2 + p^3)$ | $O(p)$ | Custo cúbico em $p$ para inverter matriz $(p \times p)$; instável com colinearidade. |
| **KNN (*k-Nearest Neighbors*)** | $O(1)$ *(lazy learner)* | $O(n \cdot p)$ por predição | Inviável para latência baixa com $n$ grande; sofre com a "maldição da dimensionalidade" (*curse of dimensionality*). |
| **Árvore de Decisão (Treino CART)** | $O(p \cdot n \log n)$ por split | $O(\text{profundidade}) \approx O(\log n)$ | Inferência ultrarrápida (poucas comparações `if/else`). |
| **Random Forest ($B$ árvores)** | $O(B \cdot p_{\text{sub}} \cdot n \log n)$ | $O(B \cdot \text{profundidade})$ | Tempo de inferência escala linearmente com o número de árvores. |
| **K-Means ($k$ clusters, $i$ iterações)** | $O(i \cdot k \cdot n \cdot p)$ | $O(k \cdot p)$ | Distância euclidiana para todos os centroides. |
| **Transformer (Atenção Padrão)** | $O(n^2 \cdot d)$ *(onde $n$ = contexto)* | $O(n^2 \cdot d)$ por token *(ou $O(n \cdot d)$ com KV-Cache)* | Gargalo quadrático no comprimento da sequência ($n$). |

A **regra assintótica** (ou simplificação assintótica) é o conjunto de regras matemáticas e conceituais usado na análise de algoritmos para determinar a sua ordem de crescimento (*Big-O*) quando o tamanho dos dados tende ao infinito ($n \to \infty$ ou $p \to \infty$).

O termo **"assintótico"** significa justamente focar no comportamento de limite de uma função, ou seja, analisar o que acontece com o tempo de execução ou com a memória quando o volume de dados se torna arbitrariamente grande.

As regras fundamentais de simplificação assintótica são:

---

### 1. Regra do Termo Dominante (ou do Maior Grau)
Quando uma função que descreve o número de operações contém múltiplos termos, apenas o termo com a maior taxa de crescimento é mantido, e todos os termos de menor ordem são descartados.

* **Exemplo matemático:**
  $$f(n) = 3n^3 + 40n^2 + 500n + 10.000$$
  À medida que $n$ cresce para valores gigantescos (ex.: $n = 10^6$), o valor de $n^3$ ($10^{18}$) é incomparavelmente maior do que $n^2$ ou $n$.  
  **Simplificação:** $O(n^3)$.

* **Exemplo em Machine Learning:**  
  Nas Equações Normais da Regressão Linear, o custo é dado por:
  $$O(n \cdot p^2 + p^3)$$
  Se o número de *features* $p$ for o foco do gargalo, a parte cúbica $p^3$ domina os termos menores relacionados à inversão matricial.

---

### 2. Regra da Constante Multiplicativa
Constantes multiplicativas (e somas constantes) são eliminadas porque elas não alteram a forma como a complexidade escala.

* **Exemplo matemático:**
  $$f(n) = 7n \implies O(n)$$
  $$f(n) = \frac{1}{2}n^2 \implies O(n^2)$$

* **Por que isso é feito?**  
  A notação assintótica não mede segundos ou ciclos de CPU exatos, mas sim a classe de escalabilidade. Um algoritmo que faz $2n$ passos e outro que faz $100n$ passos pertencem à mesma classe de complexidade linear ($O(n)$) — ambos dobram de tempo se a entrada dobrar.

---

### 3. Regra da Adição (Tarefas Sequenciais)
Se um algoritmo executa duas etapas sequenciais e independentes, a complexidade total é o teto máximo entre elas:

$$\text{Passo A } [O(f(n))] + \text{Passo B } [O(g(n))] \implies O(\max(f(n), g(n)))$$

* **Exemplo prático:**
  1. Primeiro, você faz a ordenação dos dados: $O(n \log n)$.
  2. Depois, percorre os dados com um `for` simples para aplicar uma transformação: $O(n)$.
  * **Complexidade final:** $O(n \log n)$, pois $n \log n$ domina $n$.

---

### 4. Regra da Multiplicação (Operações Aninhadas)
Se uma etapa do algoritmo é repetida internamente a cada iteração de outra etapa (laços aninhados ou chamadas iterativas), as complexidades se multiplicam:

$$\text{Loop Externo } [O(f(n))] \times \text{Loop Interno } [O(g(n))] \implies O(f(n) \cdot g(n))$$

* **Exemplo prático:**
  * Um laço percorre todas as $n$ amostras: $O(n)$.
  * Para cada amostra, ele calcula a distância euclidiana comparando suas $p$ variáveis: $O(p)$.
  * **Complexidade final:** $O(n \cdot p)$.

---

### Resumo Intuitivo
A análise assintótica serve para responder à pergunta:

> *"No pior cenário e com uma quantidade massiva de dados, quem é o verdadeiro responsável pelo estrangulamento da computação?"*

Ao aplicar a regra assintótica, eliminam-se os ruídos (constantes e termos menores) para destacar a equação matemática pura que dita se a solução continuará viável ou explodirá em produção.
